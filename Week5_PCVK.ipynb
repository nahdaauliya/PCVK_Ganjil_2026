```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

image = cv.imread('/content/drive/MyDrive/PCVK/Images (1)/Images/lena_lc.jpg')

b, g, r = cv.split(image)

def manual_equalization(channel):

    hist = np.zeros(256, dtype=np.int64)

    for pixel in channel.flatten():
        hist[pixel] += 1

    p = hist / channel.size
    cdf = np.cumsum(p)
    mapping = np.round(cdf * 255).astype(np.uint8)
    equalized = mapping[channel]
    return equalized, hist, cdf, mapping

b_equalized, hist_b, cdf_b, map_b = manual_equalization(b)
g_equalized, hist_g, cdf_g, map_g = manual_equalization(g)
r_equalized, hist_r, cdf_r, map_r = manual_equalization(r)

image_equalized = cv.merge([
    b_equalized,
    g_equalized,
    r_equalized
])

image_rgb = cv.cvtColor(image, cv.COLOR_BGR2RGB)
equalized_rgb = cv.cvtColor(image_equalized, cv.COLOR_BGR2RGB)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(image_rgb)
plt.title('Citra Sebelum Equalization')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(equalized_rgb)
plt.title('Citra Sesudah Equalization')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

path = '/content/drive/MyDrive/PCVK/Images (1)/Images/KTM Baru.jpg'

bgr = cv.imread(path)

kanal = list(cv.split(bgr))

for i in range(3):
    kanal[i] = cv.equalizeHist(kanal[i])

he_rgb = cv.merge(kanal)

ycrcb = cv.cvtColor(bgr, cv.COLOR_BGR2YCrCb)

y, cr, cb = cv.split(ycrcb)

he_y = cv.cvtColor(
    cv.merge([
        cv.equalizeHist(y),
        cr,
        cb
    ]),
    cv.COLOR_YCrCb2BGR
)

hsv = cv.cvtColor(bgr, cv.COLOR_BGR2HSV)

h, sat, v = cv.split(hsv)

he_v = cv.cvtColor(
    cv.merge([
        h,
        sat,
        cv.equalizeHist(v)
    ]),
    cv.COLOR_HSV2BGR
)

lab = cv.cvtColor(bgr, cv.COLOR_BGR2LAB)

l, a, b = cv.split(lab)

he_l = cv.cvtColor(
    cv.merge([
        cv.equalizeHist(l),
        a,
        b
    ]),
    cv.COLOR_LAB2BGR
)

judul = [
    'Asli',
    'Equalisasi B, G, R',
    'Kanal Y',
    'Kanal V',
    'Kanal L'
]

citra = [
    bgr,
    he_rgb,
    he_y,
    he_v,
    he_l
]

plt.figure(figsize=(18, 4))

for i, (c, t) in enumerate(zip(citra, judul), 1):

    plt.subplot(1, 5, i)

    plt.imshow(cv.cvtColor(c, cv.COLOR_BGR2RGB))

    plt.title(t, fontsize=9)

    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
def geser_hue(asli, hasil):

    h1 = cv.cvtColor(
        asli,
        cv.COLOR_BGR2HSV
    )[:, :, 0].astype(np.int16)

    h2 = cv.cvtColor(
        hasil,
        cv.COLOR_BGR2HSV
    )[:, :, 0].astype(np.int16)

    d = np.abs(h1 - h2)

    d = np.minimum(d, 180 - d)

    return d.mean()

print('%-22s %16s %16s' %
      ('Cara', 'Geser Hue (der)', 'Rerata terang'))

print('-' * 58)

for t, c in zip(judul, citra):

    hue_shift = geser_hue(bgr, c)

    brightness = cv.cvtColor(
        c,
        cv.COLOR_BGR2GRAY
    ).mean()

    print('%-22s %16.2f %16.2f' %
          (t, hue_shift, brightness))
```

```python
path = '/content/drive/MyDrive/PCVK/Images (1)/Images/KTM Baru.jpg'

img = cv.imread(path, cv.IMREAD_GRAYSCALE)

print("Ukuran citra:", img.shape)


hist = np.zeros(256, dtype=np.int64)

height, width = img.shape

for y in range(height):
    for x in range(width):
        hist[img[y, x]] += 1

cdf = np.cumsum(hist)
cdf_min = cdf[cdf > 0][0]

total_pixel = img.size

lut = np.round(
    (cdf - cdf_min) /
    (total_pixel - cdf_min) * 255
).clip(0, 255).astype(np.uint8)


manual_equalized = lut[img]


hist = np.zeros(256, dtype=np.int64)

height, width = img.shape

for y in range(height):
    for x in range(width):
        hist[img[y, x]] += 1

cdf = np.cumsum(hist)

cdf_min = cdf[cdf > 0][0]

total_pixel = img.size

lut = np.round(
    (cdf - cdf_min) /
    (total_pixel - cdf_min) * 255
).clip(0, 255).astype(np.uint8)

manual_equalized = lut[img]

cv_equalized = cv.equalizeHist(img)
```

```python
def hitung_psnr(asli, hasil):

    mse = np.mean(
        (asli.astype(np.float64) -
         hasil.astype(np.float64)) ** 2
    )

    if mse == 0:
        return float('inf')

    psnr = 10 * np.log10((255 ** 2) / mse)

    return psnr


psnr_manual = hitung_psnr(img, manual_equalized)
psnr_cv = hitung_psnr(img, cv_equalized)

print("PSNR manual       :", psnr_manual, "dB")
print("PSNR cv.equalizeHist :", psnr_cv, "dB")
```

PSNR antara citra asli dan citra hasil histogram equalization dapat menurun meskipun citra hasil terlihat lebih jelas. Hal ini terjadi karena PSNR mengukur kemiripan nilai piksel antara citra asli dan citra hasil. Histogram equalization memang sengaja mengubah nilai intensitas piksel untuk memperbaiki distribusi intensitas dan meningkatkan kontras. Perubahan tersebut menyebabkan MSE terhadap citra asli meningkat sehingga nilai PSNR menurun.

Dengan demikian, PSNR kurang tepat jika digunakan sebagai satu-satunya ukuran untuk menilai keterbacaan sebuah citra. Nilai PSNR yang rendah tidak selalu menunjukkan bahwa citra hasil lebih buruk, karena citra hasil dapat memiliki kontras yang lebih baik dan informasi tertentu menjadi lebih mudah dibaca. ##E-3 TUGAS PRAKTIKUM DITHERING

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

PARAM['L'] = 4


def floyd_steinberg_color(img, levels):

    work = img.astype(np.float64).copy()

    height, width, channels = work.shape

    step = 255 / (levels - 1)

    for y in range(height):
        for x in range(width):

            for c in range(channels):

                old_pixel = work[y, x, c]

                new_pixel = np.round(old_pixel / step) * step

                new_pixel = np.clip(new_pixel, 0, 255)

                error = old_pixel - new_pixel

                work[y, x, c] = new_pixel

                if x + 1 < width:
                    work[y, x + 1, c] += error * 7 / 16

                if y + 1 < height:

                    if x - 1 >= 0:
                        work[y + 1, x - 1, c] += error * 3 / 16

                    work[y + 1, x, c] += error * 5 / 16

                    if x + 1 < width:
                        work[y + 1, x + 1, c] += error * 1 / 16

    result = np.clip(work, 0, 255).astype(np.uint8)

    return result
```

```python
path_lena = '/content/drive/MyDrive/PCVK/Images (1)/Images/lena.jpg'

lena = cv.imread(path_lena)

print("Ukuran citra:", lena.shape)
print("Jumlah level:", PARAM['L'])

lena_dither = floyd_steinberg_color(
    lena,
    PARAM['L']
)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(cv.cvtColor(lena, cv.COLOR_BGR2RGB))
plt.title('Citra Asli Lena')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(cv.cvtColor(lena_dither, cv.COLOR_BGR2RGB))
plt.title(f'Floyd-Steinberg ({PARAM["L"]} level)')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
path_ktm = '/content/drive/MyDrive/PCVK/Images (1)/Images/KTM Baru.jpg'

ktm = cv.imread(path_ktm)

ktm_dither = floyd_steinberg_color(
    ktm,
    PARAM['L']
)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(cv.cvtColor(ktm, cv.COLOR_BGR2RGB))
plt.title('KTM1b - Citra Asli')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(cv.cvtColor(ktm_dither, cv.COLOR_BGR2RGB))
plt.title(f'KTM1b - Floyd-Steinberg ({PARAM["L"]} level)')
plt.axis('off')

plt.tight_layout()
plt.show()
```

###E-3 Nomor 2 - Equalization + Floyd-Steinberg

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

L = 2


def floyd_steinberg(img, levels):

    work = img.astype(np.float64).copy()

    height, width = work.shape

    step = 255 / (levels - 1)

    for y in range(height):
        for x in range(width):

            old_pixel = work[y, x]

            new_pixel = np.round(old_pixel / step) * step

            new_pixel = np.clip(new_pixel, 0, 255)

            error = old_pixel - new_pixel

            work[y, x] = new_pixel

            if x + 1 < width:
                work[y, x + 1] += error * 7 / 16

            if y + 1 < height:

                if x - 1 >= 0:
                    work[y + 1, x - 1] += error * 3 / 16

                work[y + 1, x] += error * 5 / 16

                if x + 1 < width:
                    work[y + 1, x + 1] += error * 1 / 16

    return np.clip(work, 0, 255).astype(np.uint8)
```

```python
path_lena = '/content/drive/MyDrive/PCVK/Images (1)/Images/lena_lc.jpg'

lena_lc = cv.imread(
    path_lena,
    cv.IMREAD_GRAYSCALE
)

print("Ukuran:", lena_lc.shape)

lena_dither_langsung = floyd_steinberg(
    lena_lc,
    L
)
lena_equalized = cv.equalizeHist(lena_lc)
lena_equalized_dither = floyd_steinberg(
    lena_equalized,
    L
)
```

```python
plt.figure(figsize=(16, 5))

plt.subplot(1, 3, 1)
plt.imshow(lena_lc, cmap='gray')
plt.title('Lena_lc Asli')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(lena_dither_langsung, cmap='gray')
plt.title('Dithering Langsung')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(lena_equalized_dither, cmap='gray')
plt.title('Equalisasi → Dithering')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
path_ktm = '/content/drive/MyDrive/PCVK/Images (1)/Images/KTM Baru.jpg'

ktm = cv.imread(
    path_ktm,
    cv.IMREAD_GRAYSCALE
)

ktm_dither_langsung = floyd_steinberg(
    ktm,
    L
)

ktm_equalized = cv.equalizeHist(ktm)

ktm_equalized_dither = floyd_steinberg(
    ktm_equalized,
    L
)
```

Hasil **equalization → dithering** membuat kontras citra lebih baik sehingga **NIM dan nama lebih mudah terbaca** dibandingkan dithering langsung. Hal ini karena equalization memperbaiki distribusi intensitas sebelum proses dithering.
