```python
img = cv.imread('/content/drive/MyDrive/PCVK/Images (1)/lena.jpg')
img = cv.cvtColor(img, cv.COLOR_BGR2RGB)

height, width, depth = np.shape(img)
names = np.arange(256)

red = [0]*256
green = [0]*256
blue = [0]*256

for y in range(0, height) :
  for x in range (0, width) :
    red[img[y][x][0]] += 1
    green[img[y][x][1]] += 1
    blue[img[y][x][2]] += 1


red_numpy, _ = np.histogram(img[:,:,0], bins=256, range=(0, 256))
green_numpy, _=np.histogram(img[:,:,1], bins=256, range=(0, 256))
blue_numpy,_ = np.histogram(img[:,:,2], bins=256, range=(0,256))

####################################################################

red_cv = cv.calcHist([img], [0], None, [256], [0,256]).flatten()
green_cv = cv.calcHist([img], [1], None, [256], [0,256]).flatten()
blue_cv = cv.calcHist([img], [2], None, [256], [0,256]).flatten()


######################################################################
red_cv = red_cv.astype(int)
green_cv = green_cv.astype(int)
blue_cv = blue_cv.astype(int)

max_diff_red_numpy = np.max(np.abs(np.array(red)- red_numpy))
max_diff_green_numpy = np.max(np.abs(np.array(green)- green_numpy))
max_diff_blue_numpy = np.max(np.abs(np.array(blue)- blue_numpy))

#######################################################################

max_diff_red_cv = np.max(np.abs(np.array(red)- red_cv))
max_diff_green_cv = np.max(np.abs(np.array(green)- green_cv))
max_diff_blue_cv = np.max(np.abs(np.array(blue)- blue_cv))

print("---SELISIH MANUAL DAN NUMPY---")
print("Red    :", max_diff_red_numpy)
print("Green  :", max_diff_green_numpy)
print("Blue   :", max_diff_blue_numpy)


print("---SELISIH MANUAL DAN cv.calcHist---")
print("Red    :", max_diff_red_cv)
print("Green  :", max_diff_green_cv)
print("Blue   :", max_diff_blue_cv)

#SELSISIH MAX
max_diff_np = max(
    max_diff_red_numpy,
    max_diff_green_numpy,
    max_diff_blue_numpy
)

max_diff_cv = max(
    max_diff_red_cv,
    max_diff_green_cv,
    max_diff_blue_cv
)

print("\n====HASIL AKHIR====")
print("Selisih maksimum Manual vs NumPy       :", max_diff_np)
print("Selisih maksimum Manual vs cv.calcHist :", max_diff_cv)

names =  np.arange(256)
fig, axs = plt.subplots(1, 3, figsize=[20,5], sharex= True, sharey=True)
fig.suptitle('Histogram RGB plot')
fig.text(0.09, 0.5, 'Jumlah Kemunculan', va='center', rotation='vertical')
fig.text(0.5, 0.04, 'Intensitas Warna', ha='center')
axs[0].bar(names, red, color='red')
axs[1].bar(names, green, color='green')
axs[2].bar(names, blue, color='blue')
```

```python
path = '/content/drive/MyDrive/PCVK/Images (1)/'

# Nama file
files = ['KTM Baru.jpg', 'KTM Lama.jpg']

fig, axs = plt.subplots(1,2, figsize=(16, 10))

for ax, file in zip(axs.ravel(), files):
  img = cv.imread(path + file, cv.IMREAD_GRAYSCALE)

  hist = cv.calcHist([img], [0], None, [256], [0, 256]). flatten()
  p = hist / hist.sum()
  cdf = np.cumsum(p)
  intensity = np.arange(256)

  ax.bar(intensity, p , width=1.0, alpha=0.5, label='p(j)')

  ax.plot(intensity, cdf, linewidth=2, label='CDF')

  ax.set_title(file)
  ax.set_xlabel('Intensitas')
  ax.set_ylabel('Probabilitas / CDF')
  ax.set_xlim(0, 255)
  ax.set_ylim(0, 1.05)
  ax.grid(alpha=0.3)
  ax.legend()

plt.suptitle(
    'Histogram Ternomaloasis dan CDF KTM Baru dan Lama',
    fontsize=16
)

plt.tight_layout()
plt.show()
```

Berdasarkan kurva CDF, citra yang paling gelap memiliki distribusi intensitas piksel yang lebih banyak berada pada intensitas rendah. Hal ini menyebabkan kurva CDF meningkat lebih cepat pada bagian kiri atau pada rentang intensitas yang rendah. Sebaliknya, pada citra yang paling terang, piksel lebih banyak berada pada intensitas tinggi sehingga kurva CDF mengalami kenaikan lebih banyak pada bagian kanan.

Perbedaan diatas berkaitan dengan kondisi pencahayaan saat proses akuisisi. Citra yang diambil dengan pencahayaan yang lebih rendah akan menghasilkan intensitas piksel yang cenderung rendah sehingga citra terlihat lebih gelap. Sementara itu, pencahayaan yang lebih kuat akan menghasilkan intensitas piksel yang lebih tinggi sehingga citra terlihat lebih terang.
