```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

print(' Menampilkan warna merah dan mengubah warna lain menjadi grayscale ')
print('-------------------------------------------------------------------')

# Membaca gambar
img = cv2.imread(
    '/content/drive/MyDrive/PCVK/Images/Images/peppers.jpg'
)

# Mengubah BGR menjadi HSV
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

# Membuat salinan gambar
result = img.copy()

# Akses setiap pixel
for y in range(img.shape[0]):
    for x in range(img.shape[1]):

        # Mengambil nilai HSV
        H = hsv[y, x, 0]
        S = hsv[y, x, 1]
        V = hsv[y, x, 2]

        # Mengecek warna merah
        if ((H <= 10 or H >= 170) and S > 80 and V > 50):

            # Warna merah tetap seperti asli
            result[y, x] = img[y, x]

        else:
            # Mengambil nilai BGR
            B = float(img[y, x, 0])
            G = float(img[y, x, 1])
            R = float(img[y, x, 2])

            # Mengubah menjadi grayscale
            gray = (R + G + B) / 3

            result[y, x] = [gray, gray, gray]

# Mengubah BGR menjadi RGB untuk matplotlib
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
result_rgb = cv2.cvtColor(result, cv2.COLOR_BGR2RGB)

# Menampilkan hasil
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title('Original')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(result_rgb)
plt.title('Red + Grayscale')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2 as cv
import glob
import numpy as np
import matplotlib.pyplot as plt

# Membaca semua gambar noise
cv_img = []

for img in glob.glob(
    '/content/drive/MyDrive/PCVK/Images/Images/noises/*.jpg'
):
    n = cv.imread(img)
    cv_img.append(n)

print("Jumlah gambar:", len(cv_img))


# Jumlah gambar yang akan dirata-rata
jumlah_average = [10, 20, 40, 80, 100]

# Gambar asli
original = cv.imread(
    '/content/drive/MyDrive/PCVK/Images/Images/galaxy.jpg'
)


for jumlah in jumlah_average:

    # Mengambil sejumlah gambar dan menghitung rata-rata
    hasil = np.mean(
        cv_img[:jumlah],
        axis=0
    ).astype(np.uint8)

    # Menghitung PSNR
    psnr = cv.PSNR(original, hasil)

    print(
        f"Jumlah citra: {jumlah}, PSNR: {psnr:.2f} dB"
    )

    # Menampilkan hasil
    plt.figure(figsize=(6, 4))
    plt.imshow(cv.cvtColor(hasil, cv.COLOR_BGR2RGB))
    plt.title(
        f'Average {jumlah} Citra - PSNR: {psnr:.2f} dB'
    )
    plt.axis('off')
    plt.show()
```

#### 1. Apakah peningkatan jumlah citra selalu memberikan peningkatan PSNR antar jumlah citra yang signifikan?
**Jawaban**
Tidak selalu. Penambahan jumlah citra yang sirata-ratakan umumnya dapat meningkatkan kulaitas hasil denoising dan nilai PSNR, tetapi peningkatannya semakin kecil ketika jumlah citra semakin banyak. Hal ini karena noise yang acak semakin banyak tereduksi melalui proses averaging, sehingga setelah jumlah citra tertentu, tambahan citra hanya memberikan perbaikan yang relatif kecil.

```python
import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow

print('Praktikum Image Masking dengan Berbagai Operator Logika')
print('---------------------------------------------------------')

img_path = '/content/drive/MyDrive/PCVK/Images/Images/couple.tiff'
original = cv.imread(img_path)

h, w = original.shape[:2]

mask = np.zeros((h, w), dtype=np.uint8)
cv.circle(mask, (int(w * 0.35), int(h * 0.3)), int(h * 0.25), 255, -1) # Lingkaran kiri
cv.circle(mask, (int(w * 0.65), int(h * 0.3)), int(h * 0.25), 255, -1) # Lingkaran kanan

mask_bgr = cv.cvtColor(mask, cv.COLOR_GRAY2BGR)

print("Memproses operator logika...")

# 1. OPERATOR NOT (Komplemen pada Mask)
not_mask = cv.bitwise_not(mask)
res_not = cv.bitwise_and(original, original, mask=not_mask)

# 2. OPERATOR OR
res_or = cv.bitwise_or(original, mask_bgr)

# 3. OPERATOR AND (Standard Masking)
res_and = cv.bitwise_and(original, original, mask=mask)

# 4. OPERATOR NAND (Not AND)
and_masked = cv.bitwise_and(original, original, mask=mask)
res_nand = cv.bitwise_not(and_masked)

# 5. OPERATOR XOR
res_xor = cv.bitwise_xor(original, mask_bgr)

# --- MENAMPILKAN HASIL ---
print("\n--- Menampilkan Hasil Operator ---")
print("1. Mask Utama:")
cv2_imshow(mask_bgr)

print("2. Hasil Operator AND (Masking Standar):")
cv2_imshow(res_and)

print("3. Hasil Operator NOT:")
cv2_imshow(res_not)

print("4. Hasil Operator OR:")
cv2_imshow(res_or)

print("5. Hasil Operator XOR:")
cv2_imshow(res_xor)

print("6. Hasil Operator NAND:")
cv2_imshow(res_nand)
```

Berdasarkan percobaan, setiap operator logika menghasilkan keluaran yang berbeda. Operator AND digunakan untuk melakukan image masking, yaitu mempertahankan bagian citra yang berada pada area putih mask dan menghilangkan bagian lainnya. Operator NOT membalik nilai pixel pada citra sehingga bagian yang gelap menjadi terang dan sebaliknya. Sedangkan operator OR menggabungkan nilai pixel dari citra dan mask, sehingga area mask yang bernilai putih akan memengaruhi hasil citra menjadi putih/terang. Perbedaan hasil tersebut terjadi karena setiap operator memiliki aturan operasi logika yang berbeda dalam memproses nilai pixel.

```python
import cv2
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow
import numpy as np

print(' Mengubah tingkat kecerahan menggunakan transform log ')
print('------------------------------------------------------')

try:
  brightness =int(input('Masukkan nilai kecerahan: '))
except ValueError:
  print('Error, not a number')

#sesuaikan dengan file dan folder di drive Anda
original = cv2.imread('/content/drive/MyDrive/PCVK/Images/Images/street-5963619_1280.jpg')
log_image = np.zeros(original.shape, original.dtype)

c= 255 / np.log(1+255)

#akses per piksel
#np.clip digunakan untuk melakukan truncate pixel (nilai dibatasi antara 0-255)
for y in range(original.shape[0]):
  for x in range(original.shape[1]):
    for z in range(original.shape[2]):
      log_image[y,x,z] = np.clip(original[y,x,z] + c * np.log(1+float(original[y,x,z])), 0, 255)

#cara simple tanpa for loop
#brightness_image = cv.convertScaleAbs(original, beta=brightness)

final_frame = cv2.hconcat((original, log_image))
cv2_imshow(final_frame)
```

Alasan memilih transform log:
- Foto secara keseluruhan gelap, terutama bagian pegunungan dan foreground.
- Ada beberapa bagian yang sudah punya detail/brightness cukup tinggi, seperti lampu kota, kabut, dan bulan.
- Jika hanya menaikkan brightness secara linear, bagian yang sudah terang berisiko menjadi terlalu terang/overexposure.
- Transformasi logaritmik lebih cocok untuk kondisi seperti ini karena lebih meningkatkan detail pada bagian yang gelap, sementara perubahan pada bagian yang sudah terang relatif lebih terkendali.

```python
import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow

print('Enhancing Crayfish Image (Contrast & Gamma)')
print('-------------------------------------------')

original = cv.imread('/content/drive/MyDrive/PCVK/Images/Images/crayfish.jpg')

# Apply Contrast Stretching / Linear Adjustment (Alpha = 1.5, Beta = 30)
alpha = 1.5
beta = 30
adjusted_contrast = cv.convertScaleAbs(original, alpha=alpha, beta=beta)

# Apply Gamma Correction (Gamma = 0.5 for brightening shadows)
gamma = 0.5
invGamma = 1.0 / gamma
table = np.array([((i / 255.0) ** invGamma) * 255 for i in np.arange(0, 256)]).astype("uint8")
final_enhanced = cv.LUT(adjusted_contrast, table)

# Display original and enhanced result side by side
final_frame = cv.hconcat([original, final_enhanced])
cv2_imshow(final_frame)
```

Metode Contrast digunakan untuk meningkatkan perbedaan intensitas pada citra, sedangkan Gamma Correction dengan γ < 1 digunakan untuk meningkatkan detail pada area yang gelap. Kombinasi keduanya diharapkan membuat objek utama lebih terlihat tanpa hanya menaikkan kecerahan seluruh citra.
