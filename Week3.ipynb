```python
import cv2
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow
import numpy as np

print(' Mengubah tingkat kecerahan citra ')
print('----------------------------------')

try:
  brightness =int(input('Masukkan nilai kecerahan: '))
except ValueError:
  print('Error, not a number')

#sesuaikan dengan file dan folder di drive Anda
original = cv2.imread('/content/drive/MyDrive/PCVK/Images/Images/houses.jpg')
brightness_image = np.zeros(original.shape, original.dtype)

#akses per piksel
#np.clip digunakan untuk melakukan truncate pixel (nilai dibatasi antara 0-255)
for y in range(original.shape[0]):
  for x in range(original.shape[1]):
    for c in range(original.shape[2]):
      brightness_image[y,x,c] = np.clip(original[y,x,c] + brightness, 0, 255)

#cara simple tanpa for loop
#brightness_image = cv.convertScaleAbs(original, beta=brightness)

final_frame = cv2.hconcat((original, brightness_image))
cv2_imshow(final_frame)
```

```python
import cv2
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow
import numpy as np

print(' Mengubah kontras dan tingkat kecerahan citra ')
print('----------------------------------------------')

try:
  brightness =int(input('Masukkan nilai kecerahan: '))
  contras =float(input('Masukkan nilai kontras: '))
except ValueError:
  print('Error, not a number')

original = cv2.imread('/content/drive/MyDrive/PCVK/Images/Images/sunset.jpg')
contras_image = np.zeros(original.shape, original.dtype)

#akses per piksel
#np.clip digunakan untuk melakukan truncate pixel (nilai dibatasi antara 0-255)
for y in range(original.shape[0]):
  for x in range(original.shape[1]):
    for c in range(original.shape[2]):
      contras_image[y,x,c] = np.clip(
          float(original[y,x,c]) * contras + brightness, 0, 255)

final_frame = cv2.hconcat((original, contras_image))
cv2_imshow(final_frame)
```

```python
import cv2
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow
import numpy as np

print(' Mengubah tingkat kecerahan menggunakan transform log ')
print('------------------------------------------------------')

try:
  brightness =int(input('Masukkan nilai kecerahan: '))
except ValueError:
  print('Error, not a number')

#sesuaikan dengan file dan folder di drive Anda
original = cv2.imread('/content/drive/MyDrive/PCVK/Images/Images/houses.jpg')
log_image = np.zeros(original.shape, original.dtype)

c= 255 / np.log(1+255)

#akses per piksel
#np.clip digunakan untuk melakukan truncate pixel (nilai dibatasi antara 0-255)
for y in range(original.shape[0]):
  for x in range(original.shape[1]):
    for z in range(original.shape[2]):
      log_image[y,x,z] = np.clip(original[y,x,z] + c * np.log(1+float(original[y,x,z])), 0, 255)

#cara simple tanpa for loop
#brightness_image = cv.convertScaleAbs(original, beta=brightness)

final_frame = cv2.hconcat((original, log_image))
cv2_imshow(final_frame)
```

```python
import cv2
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow
import numpy as np

print(' Mengubah image menjadi Grayscale Average ')
print('------------------------------------------')

#sesuaikan dengan file dan folder di drive Anda
img = cv2.imread('/content/drive/MyDrive/PCVK/Images/Images/peppers.jpg')
original = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
g_avg_image = np.zeros(original.shape, original.dtype)

#akses per piksel
#np.clip digunakan untuk melakukan truncate pixel (nilai dibatasi antara 0-255)
for y in range(original.shape[0]):
  for x in range(original.shape[1]):

    B = float(original[y, x, 0])
    G = float(original[y, x, 1])
    R = float(original[y, x, 2])

    g_avg_image[y,x] = np.clip((R + G + B) / 3, 0, 255)

#cara simple tanpa for loop
#brightness_image = cv.convertScaleAbs(original, beta=brightness)

final_frame = cv2.hconcat((original, g_avg_image))
plt.imshow(final_frame)
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow
import numpy as np

print(' Mengubah image menjadi Grayscale Lightness ')
print('------------------------------------------')

# try:
#   brightness =int(input('Masukkan nilai kecerahan: '))
# except ValueError:
#   print('Error, not a number')

#sesuaikan dengan file dan folder di drive Anda
img = cv2.imread('/content/drive/MyDrive/PCVK/Images/Images/peppers.jpg')
original = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
# g_avg_image = np.zeros(original.shape, original.dtype)
g_light_image = np.zeros(original.shape, original.dtype)
# g_lum_image = np.zeros(original.shape, original.dtype)

#akses per piksel
#np.clip digunakan untuk melakukan truncate pixel (nilai dibatasi antara 0-255)
for y in range(original.shape[0]):
  for x in range(original.shape[1]):

    B = float(original[y, x, 0])
    G = float(original[y, x, 1])
    R = float(original[y, x, 2])

    g_light_image[y,x] = np.clip((max(R, G, B) + min(R, G,B)) / 2, 0, 255)
    # g_lum_image[y,x] = np.clip(0.21*R + 0.72*G + 0.07*B, 0, 255)

#cara simple tanpa for loop
#brightness_image = cv.convertScaleAbs(original, beta=brightness)

final_frame = cv2.hconcat((original, g_light_image))
# final_frame2 = cv2.hconcat((original, g_lum_image))
plt.imshow(final_frame)
plt.show()
# plt.imshow(final_frame2)
```

```python
import cv2
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow
import numpy as np

print(' Mengubah image menjadi Grayscale Luminance ')
print('--------------------------------------------')

# try:
#   brightness =int(input('Masukkan nilai kecerahan: '))
# except ValueError:
#   print('Error, not a number')

#sesuaikan dengan file dan folder di drive Anda
img = cv2.imread('/content/drive/MyDrive/PCVK/Images/Images/peppers.jpg')
original = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

g_lum_image = np.zeros(original.shape, original.dtype)

#akses per piksel
#np.clip digunakan untuk melakukan truncate pixel (nilai dibatasi antara 0-255)
for y in range(original.shape[0]):
  for x in range(original.shape[1]):

    B = float(original[y, x, 0])
    G = float(original[y, x, 1])
    R = float(original[y, x, 2])

    g_lum_image[y,x] = np.clip(0.21*R + 0.72*G + 0.07*B, 0, 255)

#cara simple tanpa for loop
#brightness_image = cv.convertScaleAbs(original, beta=brightness)

final_frame = cv2.hconcat((original, g_lum_image))
plt.imshow(final_frame)
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow
import numpy as np

print(' Gamma Correction pada citra ')
print('-----------------------------')

try:
  gamma = float(input('Masukkan nilai Gamma: '))
except ValueError:
  print('Error, not a number')

original = cv2.imread('/content/drive/MyDrive/PCVK/Images/Images/houses.jpg')
gamma_image = np.zeros(original.shape, original.dtype)

for y in range(original.shape[0]):
  for x in range(original.shape[1]):
    for c in range(original.shape[2]):
      pixel = float(original[y,x,c])
      brightness_image[y,x,c] = np.clip(255 * ((pixel/255) ** gamma) + brightness, 0, 255)

final_frame = cv2.hconcat((original, brightness_image))
cv2_imshow(final_frame)
```
